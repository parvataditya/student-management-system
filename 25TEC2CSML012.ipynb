```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def factors(n):
    return [i for i in range(1, n + 1) if n % i == 0]

for num in range(1, 51):
    prime_status = "Prime" if is_prime(num) else "Not Prime"
    print(f"{num}: {prime_status}, Factors: {factors(num)}")
```

```python
#Q1. Write a Python program to create a text file containing student details

import os
import csv

FILENAME = "students.csv"
HEADERS = ["Roll Number", "Name", "Course", "Marks"]


def initialize_file():
    """Create the file with headers if it doesn't exist."""
    if not os.path.exists(FILENAME):
        with open(FILENAME, "w", newline="") as f:
            writer = csv.writer(f)
            writer.writerow(HEADERS)


def add_record():
    """Append a new student record to the CSV file."""
    roll = input("Enter Roll Number: ").strip()
    name = input("Enter Name: ").strip()
    course = input("Enter Course: ").strip()

    while True:
        marks = input("Enter Marks: ").strip()
        if marks.replace(".", "", 1).isdigit():
            break
        print("Invalid marks. Please enter a number.")

    with open(FILENAME, "a", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([roll, name, course, marks])

    print("Record added successfully.\n")


def display_records():
    """Display all student records in a formatted table."""
    if not os.path.exists(FILENAME) or os.path.getsize(FILENAME) == 0:
        print("No records found.\n")
        return

    with open(FILENAME, "r", newline="") as f:
        reader = csv.reader(f)
        rows = list(reader)

    if len(rows) <= 1:
        print("No records found.\n")
        return

    print(f"\n{'Roll No':<10}{'Name':<15}{'Course':<15}{'Marks':<10}")
    print("-" * 50)
    for row in rows[1:]:  # skip header
        roll, name, course, marks = row
        print(f"{roll:<10}{name:<15}{course:<15}{marks:<10}")
    print()


def search_record():
    """Search for a student by Roll Number."""
    if not os.path.exists(FILENAME):
        print("No records found.\n")
        return

    roll_to_find = input("Enter Roll Number to search: ").strip()

    with open(FILENAME, "r", newline="") as f:
        reader = csv.DictReader(f)
        for row in reader:
            if row["Roll Number"] == roll_to_find:
                print(f"\nRecord found:")
                print(f"Roll Number: {row['Roll Number']}")
                print(f"Name: {row['Name']}")
                print(f"Course: {row['Course']}")
                print(f"Marks: {row['Marks']}\n")
                return

    print("Record not found.\n")


def update_marks():
    """Update a student's marks by Roll Number."""
    if not os.path.exists(FILENAME):
        print("No records found.\n")
        return

    roll_to_update = input("Enter Roll Number to update: ").strip()

    with open(FILENAME, "r", newline="") as f:
        reader = csv.DictReader(f)
        rows = list(reader)
        fieldnames = reader.fieldnames

    updated = False
    for row in rows:
        if row["Roll Number"] == roll_to_update:
            while True:
                new_marks = input(f"Enter new marks for {row['Name']}: ").strip()
                if new_marks.replace(".", "", 1).isdigit():
                    break
                print("Invalid marks. Please enter a number.")
            row["Marks"] = new_marks
            updated = True
            break

    if updated:
        with open(FILENAME, "w", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(rows)
        print("Marks updated successfully.\n")
    else:
        print("Record not found.\n")


def main():
    initialize_file()

    while True:
        print("===== Student Record Management =====")
        print("1. Add New Record")
        print("2. Display All Records")
        print("3. Search by Roll Number")
        print("4. Update Marks")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            add_record()
        elif choice == "2":
            display_records()
        elif choice == "3":
            search_record()
        elif choice == "4":
            update_marks()
        elif choice == "5":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.\n")


if __name__ == "__main__":
    main()
```

```python
#Q2. Write a Python program to read a text file and perform the following
import csv

import os
from google.colab import files

uploaded = files.upload()


FILENAME = "/content/" + list(uploaded.keys())[0]
OUTPUT_FILE = "analysis.txt"

def analyze_file():
    """Read and analyze student records from CSV file."""

    if not os.path.exists(FILENAME):
        print("students.csv file not found.")
        return

    # Read the entire file as raw text first (for line counting)
    with open(FILENAME, "r") as f:
        raw_lines = f.readlines()

    if len(raw_lines) == 0:
        print("File is empty.")
        return

    # Number of lines = total lines in the file, including header
    number_of_lines = len(raw_lines)

    # Read records using DictReader for field-based analysis
    with open(FILENAME, "r", newline="") as f:
        reader = csv.DictReader(f)
        records = list(reader)
        headers = reader.fieldnames

    if len(records) == 0:
        print("No student records found (only header present).")
        return

    # Build a list of actual WORDS (not CSV fields) by splitting on whitespace
    # Includes header row so the whole file is genuinely analyzed
    words = []
    words.extend(headers)  # header words, e.g. "Roll", "Number" if split by space

    for record in records:
        for field in [record["Roll Number"], record["Name"], record["Course"], record["Marks"]]:
            words.extend(field.split())  # splits multi-word names like "Rahul Sharma" properly

    # Count words
    number_of_words = len(words)

    # Count characters (sum of all word characters, no spaces/commas)
    number_of_characters = sum(len(word) for word in words)

    # Search for a particular word
    search_word = input("Enter the word to search: ").strip()

    occurrence = sum(1 for word in words if word.lower() == search_word.lower())

    # Find longest and shortest words
    longest_word = max(words, key=len)
    shortest_word = min(words, key=len)

    # Display results
    print("\n===== Student CSV Analysis =====")
    print("Number of lines:", number_of_lines)
    print("Number of words:", number_of_words)
    print("Number of characters:", number_of_characters)
    print(f"Occurrence of '{search_word}':", occurrence)
    print("Longest word:", longest_word)
    print("Shortest word:", shortest_word)

    # Store results in a separate TEXT file
    with open(OUTPUT_FILE, "w") as f:
        f.write("===== Student CSV Analysis =====\n")
        f.write(f"Number of lines: {number_of_lines}\n")
        f.write(f"Number of words: {number_of_words}\n")
        f.write(f"Number of characters: {number_of_characters}\n")
        f.write(f"Occurrence of '{search_word}': {occurrence}\n")
        f.write(f"Longest word: {longest_word}\n")
        f.write(f"Shortest word: {shortest_word}\n")

    print(f"\nAnalysis results saved in {OUTPUT_FILE}")


# Call the function
analyze_file()
```

```python
#Q3. Create a Python program to store employee information
import json
import os
from google.colab import files

uploaded = files.upload()


FILENAME = "/content/" + list(uploaded.keys())[0]
def initialize_file():
    """Create the JSON file with an empty list if it doesn't exist."""
    if not os.path.exists(FILENAME):
        with open(FILENAME, "w") as f:
            json.dump([], f)


def load_data():
    """Load all employee records from the JSON file."""
    with open(FILENAME, "r") as f:
        return json.load(f)


def save_data(data):
    """Save the employee list back to the JSON file."""
    with open(FILENAME, "w") as f:
        json.dump(data, f, indent=4)


def add_employee():
    """Add a new employee record."""
    data = load_data()

    emp_id = input("Enter Employee ID: ").strip()

    # Check for duplicate ID
    for emp in data:
        if emp["Employee ID"] == emp_id:
            print("Employee ID already exists. Cannot add duplicate.\n")
            return

    name = input("Enter Name: ").strip()
    department = input("Enter Department: ").strip()
    designation = input("Enter Designation: ").strip()

    while True:
        salary = input("Enter Salary: ").strip()
        if salary.replace(".", "", 1).isdigit():
            break
        print("Invalid salary. Please enter a number.")

    employee = {
        "Employee ID": emp_id,
        "Name": name,
        "Department": department,
        "Designation": designation,
        "Salary": float(salary)
    }

    data.append(employee)
    save_data(data)
    print("Employee record added successfully.\n")


def display_employees():
    """Display all employee records."""
    data = load_data()

    if not data:
        print("No employee records found.\n")
        return

    print(f"\n{'ID':<10}{'Name':<15}{'Department':<15}{'Designation':<15}{'Salary':<10}")
    print("-" * 65)
    for emp in data:
        print(f"{emp['Employee ID']:<10}{emp['Name']:<15}{emp['Department']:<15}"
              f"{emp['Designation']:<15}{emp['Salary']:<10}")
    print()


def search_employee():
    """Search for an employee by Employee ID."""
    data = load_data()
    emp_id = input("Enter Employee ID to search: ").strip()

    for emp in data:
        if emp["Employee ID"] == emp_id:
            print("\nEmployee found:")
            print(f"Employee ID: {emp['Employee ID']}")
            print(f"Name: {emp['Name']}")
            print(f"Department: {emp['Department']}")
            print(f"Designation: {emp['Designation']}")
            print(f"Salary: {emp['Salary']}\n")
            return

    print("Employee not found.\n")


def update_employee():
    """Update an existing employee's details."""
    data = load_data()
    emp_id = input("Enter Employee ID to update: ").strip()

    for emp in data:
        if emp["Employee ID"] == emp_id:
            print(f"Updating record for {emp['Name']}. Leave blank to keep current value.")

            new_name = input(f"Name [{emp['Name']}]: ").strip()
            new_department = input(f"Department [{emp['Department']}]: ").strip()
            new_designation = input(f"Designation [{emp['Designation']}]: ").strip()
            new_salary = input(f"Salary [{emp['Salary']}]: ").strip()

            if new_name:
                emp["Name"] = new_name
            if new_department:
                emp["Department"] = new_department
            if new_designation:
                emp["Designation"] = new_designation
            if new_salary:
                if new_salary.replace(".", "", 1).isdigit():
                    emp["Salary"] = float(new_salary)
                else:
                    print("Invalid salary entered. Salary not updated.")

            save_data(data)
            print("Employee record updated successfully.\n")
            return

    print("Employee not found.\n")


def delete_employee():
    """Delete an employee record by Employee ID."""
    data = load_data()
    emp_id = input("Enter Employee ID to delete: ").strip()

    for emp in data:
        if emp["Employee ID"] == emp_id:
            confirm = input(f"Are you sure you want to delete {emp['Name']}'s record? (y/n): ").strip().lower()
            if confirm == "y":
                data.remove(emp)
                save_data(data)
                print("Employee record deleted successfully.\n")
            else:
                print("Deletion cancelled.\n")
            return

    print("Employee not found.\n")


def main():
    initialize_file()

    while True:
        print("===== Employee Record Management =====")
        print("1. Add New Employee")
        print("2. Display All Employees")
        print("3. Search Employee by ID")
        print("4. Update Employee Record")
        print("5. Delete Employee Record")
        print("6. Exit")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == "1":
            add_employee()
        elif choice == "2":
            display_employees()
        elif choice == "3":
            search_employee()
        elif choice == "4":
            update_employee()
        elif choice == "5":
            delete_employee()
        elif choice == "6":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.\n")


if __name__ == "__main__":
    main()
```

```python
#Q4. Create a text file containing student records in the following format:
import csv
import json
import os


CSV_FILE = "students.csv"
JSON_FILE = "students.json"


def read_csv_file():
    """Read student records from the CSV file."""
    if not os.path.exists(CSV_FILE):
        print(f"{CSV_FILE} not found.")
        return None

    with open(CSV_FILE, "r", newline="") as f:
        reader = csv.reader(f)
        rows = list(reader)

    return rows


def convert_to_dict_list(rows):
    """Convert each row into a dictionary and build a list of dictionaries."""
    students = []

    for row in rows:
        if not row:
            continue  # skip empty lines

        roll, name, course, marks = row

        student = {
            "Roll Number": roll.strip(),
            "Name": name.strip(),
            "Course": course.strip(),
            "Marks": int(marks.strip())
        }
        students.append(student)

    return students


def save_to_json(students):
    """Store the list of dictionaries into a JSON file."""
    with open(JSON_FILE, "w") as f:
        json.dump(students, f, indent=4)

    print(f"Data successfully saved to {JSON_FILE}\n")


def display_from_json():
    """Read the JSON file and display records in a formatted manner."""
    if not os.path.exists(JSON_FILE):
        print(f"{JSON_FILE} not found.")
        return

    with open(JSON_FILE, "r") as f:
        students = json.load(f)

    if not students:
        print("No records found in JSON file.\n")
        return

    print(f"\n{'Roll No':<10}{'Name':<12}{'Course':<10}{'Marks':<8}")
    print("-" * 40)
    for student in students:
        print(f"{student['Roll Number']:<10}{student['Name']:<12}"
              f"{student['Course']:<10}{student['Marks']:<8}")
    print()


def main():
    # Step 1: Read from CSV file
    rows = read_csv_file()
    if rows is None:
        return

    # Step 2: Convert to list of dictionaries
    students = convert_to_dict_list(rows)

    # Step 3: Store in JSON file
    save_to_json(students)

    # Step 4: Read JSON file and display formatted records
    display_from_json()


if __name__ == "__main__":
    main()
```

```python
#Q5. Write a Python program to demonstrate the basic file-handling operations
import os


FILENAME = "demo.txt"


def demonstrate_file_handling():

    # ===================================================
    # 1. CREATE AND OPEN A FILE USING open(), MODE 'w'
    # ===================================================
    print("===== 1. Creating and opening file in 'w' mode =====")
    file = open(FILENAME, "w")
    print(f"File '{FILENAME}' created/opened in write mode.\n")
    file.close()  # closing manually here to demonstrate close()

    # ===================================================
    # 2. WRITE MODE ('w') — using write() and writelines()
    # ===================================================
    print("===== 2. Writing data using write() and writelines() =====")
    file = open(FILENAME, "w")  # 'w' overwrites existing content
    file.write("This is the first line.\n")
    file.write("This is the second line.\n")

    lines_to_add = ["This is the third line.\n", "This is the fourth line.\n"]
    file.writelines(lines_to_add)

    file.close()
    print("Data written successfully using write() and writelines().\n")

    # ===================================================
    # 3. READ MODE ('r') — using read(), readline(), readlines()
    # ===================================================
    print("===== 3. Reading data using read() =====")
    file = open(FILENAME, "r")
    content = file.read()
    print("Full content using read():\n" + content)
    file.close()

    print("===== Reading data using readline() =====")
    file = open(FILENAME, "r")
    first_line = file.readline()
    second_line = file.readline()
    print("First line:", first_line.strip())
    print("Second line:", second_line.strip())
    file.close()

    print("\n===== Reading data using readlines() =====")
    file = open(FILENAME, "r")
    all_lines = file.readlines()
    print("All lines as a list:", all_lines)
    file.close()

    # ===================================================
    # 4. APPEND MODE ('a') — adding data without overwriting
    # ===================================================
    print("\n===== 4. Appending data using 'a' mode =====")
    file = open(FILENAME, "a")
    file.write("This is an appended line.\n")
    file.close()

    file = open(FILENAME, "r")
    print("Content after appending:\n" + file.read())
    file.close()

    # ===================================================
    # 5. READ + WRITE MODE ('r+') — read and modify together
    # ===================================================
    print("===== 5. Using 'r+' mode to read and write =====")
    file = open(FILENAME, "r+")
    print("Content before modification:\n" + file.read())

    file.write("This line was added using r+ mode.\n")
    file.close()

    file = open(FILENAME, "r")
    print("Content after r+ modification:\n" + file.read())
    file.close()

    # ===================================================
    # 6. seek() AND tell()
    # ===================================================
    print("===== 6. Demonstrating seek() and tell() =====")
    file = open(FILENAME, "r")

    print("Current cursor position (start):", file.tell())

    data_chunk = file.read(10)  # read first 10 characters
    print("First 10 characters:", data_chunk)
    print("Cursor position after reading 10 chars:", file.tell())

    file.seek(0)  # move cursor back to the beginning
    print("Cursor position after seek(0):", file.tell())

    print("Re-reading from start:", file.read(10))
    file.close()

    # ===================================================
    # 7. USING with open() — AUTOMATIC FILE CLOSING
    # ===================================================
    print("\n===== 7. Using 'with open()' statement =====")
    with open(FILENAME, "r") as file:
        content = file.read()
        print("Content read using 'with open()':\n" + content)
    # File is automatically closed here — no need to call file.close()
    print("File automatically closed after 'with' block.")
    print("Is file closed?", file.closed)


if __name__ == "__main__":
    demonstrate_file_handling()
```
